```python
import numpy as np
import pandas as pd
```

```python
employees = {
    "ID": [1, 2, 3, 4, 5, 6],
    "Name": ["Anna", "Ben", "Sara", "David", "Emma", "Leo"],
    "Department": ["IT", "Sales", "IT", "Sales", "HR", "IT"],
    "Salary": [48000, 52000, 55000, 50000, 46000, 60000]
}

df = pd.DataFrame(employees)
```

##### Average salary by department

```python
average_salary_by_department = df.groupby("Department")["Salary"].mean()
print(f"Average salary per department is: {average_salary_by_department.round(2)}")
```

##### Maximum salary by department

```python
max_salary_by_department = df.groupby("Department")["Salary"]. max()
print(f"Maximum salary per department is: {max_salary_by_department.round(2)}")
```

##### Number of employees in each department



```python
num_of_employees_per_department = df["Department"].value_counts()
#or num_employee_per_department = df.groupby("Department")["Department"].count()

print(f"Number of employees per department is: {num_of_employees_per_department.round(2)}")
```

##### Mean, minimum and maximum salary together


```python
df2 = df["Salary"].agg(["min", "max", "mean"])
print(f"The values are: {df2.round(2)}")
```

##### Department with the highest average salary

```python
print(df.loc[df["Salary"].idxmax(), "Department"])
```

## Practice

#### Create a second DataFrame:

```python
performance = {
    "ID": [1, 2, 3, 4, 6],
    "Score": [98, 78, 92, 74, 95]
}

df_performance = pd.DataFrame(performance)
```

```python
pd.merge(df, df_performance, on="ID")
```

#### Merge employee and performance data while keeping all employees.

```python
df_new = pd.merge(df, df_performance, on="ID", how="left")
print(df_new)
```

#### Find employees whose salary is above €50,000.


```python
df2 = df_new[df_new["Salary"]>50000]
print(df2)
```

#### Find the average salary of each department.


```python
average_salary_per_department = df_new.groupby("Department")["Salary"].mean()
print(f"The average salary of each department is: {average_salary_per_department.round(2)}")
```

#### Find the average performance score of each department.

```python
average_performance_score_per_department = df_new.groupby("Department")["Score"].mean()
print(f"The average performance score of each department is: {average_performance_score_per_department.round(2)}")
```

#### Find the employee with the highest salary.


```python
employee_with_highest_salary = df_new.loc[df_new["Salary"].idxmax()]
print(employee_with_highest_salary)
```

#### Find the employee with the highest performance score.


```python
employee_with_highest_performance_score = df_new.loc[df_new["Score"].idxmax()]
print(employee_with_highest_performance_score)
```

```python
df_new["Licence"] = ["Bachelor", "Master", "PhD", "Ausbildung", "Bachelor", "Master"]
print(df_new)
```
