```python
import pandas as pd
```

## Create a dataset

```python
data = {
    "Name": ["Alice", "Sara", "Bob", "David"],
    "Age": [20, 22, 30, 18],
    "Score": [88, 65, 59, 78]
}
df = pd.DataFrame(data)
```

### Display first rows

```python
print(df.head(2))
```

### Display last rows

```python
print(df.tail(2))
```

### Shape of the dataset

```python
print(df.shape)
```

### column names:

```python
# print(df.head(0))
```

```python
print(df.columns)
```

### Print the Names

```python
print(df.loc[:, "Name"])
```

### Print the Names and Scores

```python
print(df.loc[:, ["Name", "Score"]])
```

### Print students with Age >= 22

```python
print(df[df["Age"] >= 22])
```

### Print students with Score >= 75

```python
print(df[df["Score"] >= 75])
```

### Sort by Age and Score descending:

```python
df_age_sorted = df.sort_values(by="Age", ascending=False)
print(df_age_sorted)
```

```python
df_score_sorted = df.sort_values(by="Score", ascending=False)
print(df_score_sorted)
```

### Add columns with conditions

```python
## Scores more than 75 are passed ##
df["Passed"] = df["Score"] > 75
print(df)
```

```python
## Add a bouns score of 5 ##
df["Updated_Score"] = df["Score"] + 5
print(df)
```

## Challenge: 

```python
## Read the CSV. ##
df_csv = pd.read_csv("students.csv")
print(df_csv)
```

```python
## Keep only students with Score > 80. ##
df_csv = df_csv[df_csv["Score"] > 80]
print(df_csv)
```

```python
## Sort them by Score (highest first). ##
df_csv_sorted = df_csv.sort_values(by="Score", ascending=False)
print(df_csv_sorted)
```

```python
## Add a column ##
df_csv["Excellent"] = df_csv["Score"] >= 90
print(df_csv)
```

```python
## Save a new CSV ##
df_csv.to_csv("top_students.csv", index=False)
```
