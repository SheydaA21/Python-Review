```python
## Define the student's features ##
student = {
    "name" : "Sahar",
    "age" : 35,
    "country" : "Iran",
    "major" : "Medical Engineering",
    "semester No." : 3
}

## Ask for the score ##
print("Please insert the score:")
score = float(input())

## Replace the name with upper_case letters ##
student["name"] = student["name"].upper()

if score >= 10 :
    print(f"{student['name']} has passed the exam with a score of {score}")
else: print("The student failed")
```
