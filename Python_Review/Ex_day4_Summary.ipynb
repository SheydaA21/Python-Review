```python
## Class definition
import math
class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi*self.radius**2

    def circumference(self):
        return 2*math.pi*self.radius


    
```

```python
circle1 = Circle(5)
print(f"The area is {circle1.area()}")
print(f"The circumference is {circle1.circumference()}")
```
